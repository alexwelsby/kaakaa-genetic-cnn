```python
import os
import shutil

source_dir = "./"
# assuming we're in cgp-cnn rn
target_dir = "../drive/MyDrive/cgp-cnn-TRASH"
#make the folder on my drive if it doesn't already exist
os.makedirs(target_dir, exist_ok=True)

#keeping everything that was already in the cgp-cnn directory
keep_files = [
    "cnn_train.py",
    "kaakaa_dataset.py",
    "LICENSE",
    "README.md",
    "requirements.txt",
    "cgp_config.py",
    "cgp.py",
    "cnn_model.py",
    "exp_main.py"
]

#moving our log and pickles
for f in os.listdir(source_dir):
    f_path = os.path.join(source_dir, f)
    if f not in keep_files:
        shutil.move(f_path, target_dir)
        print(f"Moved {f} -> {target_dir}")
```
