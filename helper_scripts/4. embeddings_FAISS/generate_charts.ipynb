```python
def generate_chart_for_dataset(dataset_name):
  import plotly.graph_objects as go

  categories = list(DINOV2_RESULTS[dataset_name].keys())

  #print(f"DINO len {len(DINOV2_RESULTS[dataset_name])}")
  #print(f"CNN len {len(CNN_RESULTS[dataset_name])}")

  dino_accuracies = {}
  cnn_accuracies = {}
  for category in categories:
    dino_accuracies[category] = DINOV2_RESULTS[dataset_name][category]['acc']
    total_effort = DINOV2_RESULTS[dataset_name][category]['correct'] + DINOV2_RESULTS[dataset_name][category]['incorrect']
    cnn_accuracies[category] = CNN_RESULTS[dataset_name][category]['acc']
    index = categories.index(category)
    categories[index] = str(f"{category} ({total_effort})")
    #print(category)

  #print(cnn_accuracies)
  #print(dino_accuracies)

  fig = go.Figure()


  fig.add_trace(go.Scatterpolar(
      r=list(cnn_accuracies.values()),
      theta=categories,
      #text=[f"{v * 100:.1f}%" for v in cnn_accuracies.values()],
      fill='toself',
      name='CNN Accuracy',
      textposition='top center',
      mode='lines+markers+text',
      textfont=dict(
        color='blue',
        size=12,
        family='Arial'
    )
  ))

  fig.add_trace(go.Scatterpolar(
      r=list(dino_accuracies.values()),
      theta=categories,
      #text=[f"{v * 100:.1f}%" for v in dino_accuracies.values()],
      fill='toself',
      name='DINOV2 Accuracy',
      textposition='bottom left',
      mode='lines+markers+text',
      textfont=dict(
        color='red',
    )

))

  fig.update_layout(
    title=dict(
        text=f"DINOv2 vs CNN Accuracy Comparison for Dataset {dataset_name}",
        x=0.5,
        y=0.95,
        xanchor='center',
        yanchor='top',
        font=dict(size=24, family='Arial Black')
    ),

    polar=dict(
      radialaxis=dict(
        visible=True,
        range=[0, 1]
      ),
      angularaxis=dict(
            tickfont=dict(size=20, family='Arial')
        ),),
  showlegend=True,
  legend=dict(
        font=dict(size=16, family='Arial'),
        title=dict(
            text="<b>Model</b>",
            font=dict(size=16, family='Arial')
        ),
        orientation='h',                           # Optional: horizontal legend
        yanchor='bottom',
        y=1.1,
        xanchor='center',
        x=0.5
    ),
  width=750,   # Adjust figure width
  height=600,  # Adjust figure height
  margin=dict(l=80, r=80, t=150, b=50)
)

  fig.show()

generate_chart_for_dataset('A')
```
